```python
import pygame
import cv2
import numpy as np

print(f"Cv2 Version = {cv2.__version__}")

pygame.init()

width = 1280
height = 720
window = pygame.display.set_mode((width, height))
pygame.display.set_caption("Opencv Integration")
fps = 35
clock = pygame.time.Clock()

def Webcam(cameras : int):
    if cameras and type(cameras) == int:
        return cv2.VideoCapture(cameras)

webcam = Webcam(0)
webcam.set(3,1280)
webcam.set(4,720)

loop = True
while loop:
    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            loop = False
            pygame.quit()

    red, green, blue, orange , pink = (255,0,0), (0,255,0), (0,0,255), (235, 178, 94), (201, 86, 227)
    success, img = webcam.read()
    imgRGB = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    imgRGB = np.rot90(imgRGB)
    frame = pygame.surfarray.make_surface(imgRGB).convert()
    frame = pygame.transform.flip(frame, True, False)
    window.blit(frame, (0,0))

    pygame.display.update()       
    clock.tick(fps)
```
