```python
import pygame
pygame.init()
import os

script_dir = os.path.dirname(os.path.abspath(os.getcwd()))
font_path = os.path.join(script_dir, "Resources", "Bold.tff")
balloon_path = os.path.join(script_dir, "Resources", "Balloon.png")
pizza_path = os.path.join(script_dir, "Resources", "Pizza.png")

balloon = pygame.image.load(balloon_path).convert_alpha()
pizza = pygame.image.load(pizza_path).convert_alpha()

width = 1280
height = 720
window = pygame.display.set_mode((width, height))
pygame.display.set_caption("Text & Images")
fps = 35
clock = pygame.time.Clock()

loop = True
while loop:
    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            loop = False
            pygame.quit()

    red, green, blue, orange , pink = (255,0,0), (0,255,0), (0,0,255), (235, 178, 94), (201, 86, 227)
    window.fill((235, 230, 216)) 

    font = pygame.font.Font(None, 100)
    font2 = pygame.font.Font(name=font_path, size=80)
    text = font.render("Hello! I love Python", True, (50,50,50))
    text2 = font2.render("I am Studying Computer Vision", True, (102,103,104))
    window.blit(text, (350,300))
    window.blit(text2, (100,50))

    window.blit(balloon, (210,200))
    window.blit(pizza, (50,60))
    
    pygame.display.update()       
    clock.tick(fps)
```
