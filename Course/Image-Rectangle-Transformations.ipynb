```python
import pygame
pygame.init()
import os

width = 1280
height = 720
window = pygame.display.set_mode((width, height))
pygame.display.set_caption("Images & Rectangle Transformations")
fps = 35
clock = pygame.time.Clock()

script_dir = os.path.dirname(os.path.abspath(os.getcwd()))
balloon_path = os.path.join(script_dir, "Resources", "Balloon.png")
pizza_path = os.path.join(script_dir, "Resources", "Pizza.png")

balloon = pygame.image.load(balloon_path).convert_alpha()
pizza = pygame.image.load(pizza_path).convert_alpha()
rect_balloon = balloon.get_rect()
rectNew = pygame.Rect(500,0,200,500)

balloon_rotated, pizza_rotated = pygame.transform.rotate(balloon, 90), pygame.transform.rotate(pizza, 90)
balloon_grayscaled, pizza_grayscaled = pygame.transform.grayscale(balloon),  pygame.transform.grayscale(pizza),
balloon_flipped, pizza_flipped = pygame.transform.flip(balloon, flip_x=True, flip_y=False), pygame.transform.flip(pizza, flip_x=False, flip_y=True)

loop = True
while loop:
    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            loop = False
            pygame.quit()

    red, green, blue, orange , pink = (255,0,0), (0,255,0), (0,0,255), (235, 178, 94), (201, 86, 227)
    window.fill((215, 229, 252)) 

    rect_balloon.x += 2
    print(f"Is our Balloon Hitting the Rectange = {rect_balloon.colliderect(rectNew)}")

    window.blit(balloon, (50, 100))
    window.blit(balloon_flipped, (350, 100))
    window.blit(balloon_grayscaled, (650, 100))
    window.blit(balloon_rotated, (950, 100))

    window.blit(pizza, (50, 450))
    window.blit(pizza_flipped, (350, 450))
    window.blit(pizza_grayscaled, (650, 450))
    window.blit(pizza_rotated, (950, 450))
    
    pygame.draw.rect(window, (pink), rectNew)
    window.blit(balloon, rect_balloon)

    pygame.display.update()       
    clock.tick(fps)
```
