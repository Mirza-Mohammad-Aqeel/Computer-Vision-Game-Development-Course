```python
import pygame
pygame.init()

width = 1280
height = 720
window = pygame.display.set_mode((width, height))
pygame.display.set_caption("My Pygame Template")
fps = 35
clock = pygame.time.Clock()

loop = True
while loop:
    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            loop = False
            pygame.quit()

    red, green, blue, orange , pink = (255,0,0), (0,255,0), (0,0,255), (235, 178, 94), (201, 86, 227)
    window.fill((235, 230, 216)) 

    pygame.draw.polygon(surface=window,points=((491,100),(788,100),(937,357), (788,614), (491,614), (342,357)), color=red)
    pygame.draw.circle(surface=window,color=green,center=(640,360),radius=180)
    pygame.draw.line(window, blue, (468,392), (812,392), width=10)
    pygame.draw.rect(window, orange, (468,307, 345,70), border_radius=50)
    pygame.display.update()       
    
    clock.tick(fps)
```
